```python
from faker import Faker
import random
import openpyxl
import pandas as pd
```

```python
# Defino la instancia fake
fake = Faker('ES_ar')
```

```python
# Creo el libro
wb = openpyxl.Workbook()

# Ingreso al archivo
ws = wb.active

# Defino el título
ws.title = 'Zurich'
```

```python
# Posibilidades
ramos = ['Moto',"Auto" ,"Monopatín"]
estado_pago = ['Pagado', 'Pendiente']
forma_pago = ['Mensual', 'Anual', 'Semestral']
tipo_comision = ['Directa', 'Indirecta', 'Bonificada']
```

```python
# Creo los nombres de las columnas
ws.append(["Seguro","FechaReporte", "Producto", "NoPóliza", "Prima", "Comisión (%)", 
"MontoComisión", "EstadoPago", "Vendedor"])
```

```python
# Creo las filas para la sheet "Zurich"
for i in range(4000):
    seguro = "Zurich"
    fecha = fake.date_between(start_date='-6M', end_date='today')
    producto = random.choice(ramos)

    if random.random() < 0.1:
        poliza = None
    else: 
        poliza = fake.bothify(text='??-#####') 

    if random.random() < 0.1:
        prima = -round(random.uniform(10000, 500000), 2)
    else: 
        prima = round(random.uniform(10000, 500000), 2)
    
    if random.random() < 0.1:
        pct_comision = random.choice([150, -10, 0])
    else:
        pct_comision = random.choice([5, 10, 12.5])

    comision = round(prima * pct_comision / 100, 2)

    if random.random() < 0.1:
        estado = None
    else:
        estado = random.choice(estado_pago)

    vendedor = fake.name()

    ws.append([seguro, fecha.strftime("%d/%m/%Y"), producto, poliza, prima, pct_comision, 
    estado, vendedor])

    if random.random() < 0.1:
        ws.append([seguro, fecha.strftime("%d/%m/%Y"), producto, poliza, prima, pct_comision, 
    estado, vendedor])
```

```python
# Creo el nombre de la segunda hoja
ws2 = wb.create_sheet("Sancor")
```

```python
# Creo el nombre de las columnas de la segunda hoja
ws2.append(["empresa","FechaPago", "TipoSeguro", "N° Póliza", "Prima ($)", "Comisión (%)", "Comision ($)", 
"PagoEstado", "AgenteComercial", "Productor"])
```

```python
# Creo las filas para la sheet "Sancor"
for i in range(5000):
    empresa = "Sancor"
    fecha = fake.date_between(start_date='-6M', end_date='today')
    tipo_seguro = random.choice(ramos)
    poliza = fake.bothify(text='??-#####')
    prima = round(random.uniform(10000, 500000), 2)
    pct_comision = random.choice([5, 10, 12.5])
    comision_monto = round(prima * pct_comision / 100, 2)
    pago_estado = random.choice(estado_pago)
    agente_comercial = fake.name()
    productor = fake.name()
    
    # ERROR 1: Formato de fecha incorrecto
    if random.random() < 0.05:
        fecha_str = fecha.strftime("%d/%m/%Y")  # formato europeo
    else:
        fecha_str = fecha.strftime("%Y-%m-%d")  # formato correcto

    # ERROR 2: Comisión faltante
    if random.random() < 0.03:
        pct_comision = None

    # ERROR 3: Comisión como texto
    if random.random() < 0.02:
        pct_comision = "diez"

    # ERROR 4: Prima negativa
    if random.random() < 0.01:
        prima = -prima

    # ERROR 5: Caracteres extraños en nombres
    if random.random() < 0.02:
        agente_comercial += " @#$%"
    if random.random() < 0.02:
        productor += " ***"

    fila = [empresa, fecha_str, tipo_seguro, poliza, prima, pct_comision,
            comision_monto, pago_estado, agente_comercial, productor]

    # Agrego la fila una vez
    ws2.append(fila)

    # ERROR 6: Duplicado
    if random.random() < 0.1:
        ws2.append(fila)
```

```python
# Creo la tercer hoja
ws3 = wb.create_sheet("Mapfre")
```

```python
# Creo los encabezados para la hoja 3
ws3.append(["Compañía","FechaCreación", "Producto", "Póliza"
            , "PrimaTotal", "Comisión(%)", 
            "MontoComisión", "tipo_comision", "EstadoDePago"])
```

```python
for i in range(3000):
    compañia = random.choice(["MpaFre", "Mapre", "MapFre", "Mapfre"])  # error tipográfico intencional
    fecha = fake.date_between(start_date='-6M', end_date='today')
    producto = random.choice(ramos + ["", "XXX"])  # producto inválido o vacío
    poliza = fake.bothify(text='??-#####')
    prima = round(random.uniform(10000, 500000), 2)

    # ERROR 1: Porcentaje de comisión inválido (mayor al 100 o negativo)
    if random.random() < 0.03:
        pct_comision = random.choice([-5, 105])
    else:
        pct_comision = random.choice([5, 10, 12.5])

    # ERROR 2: Monto de comisión con desvío (no coincide con prima * %)
    if random.random() < 0.05:
        monto_comision = round(random.uniform(1000, 10000), 2)  # monto erróneo
    else:
        monto_comision = round(prima * pct_comision / 100, 2)

    tipo_comision_actual = random.choice(tipo_comision)

    # ERROR 3: Estado de pago inválido
    estado = random.choice(estado_pago + ["NoPagado", "X", "—"])

    fila = [
        compañia,
        fecha.strftime("%Y-%m-%d"),
        producto,
        poliza,
        prima,
        pct_comision,
        monto_comision,
        tipo_comision_actual,
        estado
    ]

    ws3.append(fila)

    # ERROR 4: Duplicado
    if random.random() < 0.07:
        ws3.append(fila)
```

```python
# Crear la cuarta hoja
ws4 = wb.create_sheet("La Caja")
```

```python
# Creo los encabezados para "La Caja"
ws4.append(["Aseguradoras", "Fecha", "TipoSeguro","CódigoPóliza",
            "Prima","Comisión($)", "Forma_Pago", 
            "EstadoPago"])
```

```python
for i in range(5000):
    # Error tipográfico en nombre de aseguradora
    aseguradora = random.choice(["La Caja", "Caja", "LaCja", "La Caja de Ahorro"])

    # Error de fecha con formato alternativo o inválido
    if random.random() < 0.03:
        fecha = fake.date().replace("-", "/")  # formato no esperado
    else:
        fecha = fake.date_between(start_date='-6M', end_date='today').strftime("%d/%m/%Y")

    # Producto inválido o vacío
    tipo_seguro = random.choice(ramos + ["", "###", None])

    poliza = fake.bothify(text='??-#####')

    # Prima errónea (negativa o cero)
    if random.random() < 0.02:
        prima = random.choice([0, -10000])
    else:
        prima = round(random.uniform(10000, 500000), 2)

    pct_comision = random.choice([5, 10, 12.5])

    # Monto de comisión incorrecto
    if random.random() < 0.05:
        comision = round(random.uniform(1000, 20000), 2)
    else:
        comision = round(prima * (pct_comision / 100), 2)

    # Forma de pago con errores
    forma_pago_valor = random.choice(forma_pago + ["", "XXX", None])

    # Estado pago inválido
    estado_pago_valor = random.choice(estado_pago + ["Error", "Pagao", "—"])

    fila = [aseguradora, fecha, tipo_seguro, poliza, prima, comision, forma_pago_valor, estado_pago_valor]
    ws4.append(fila)

    # Agregar duplicados
    if random.random() < 0.06:
        ws4.append(fila)
```

```python
# Creo encabezados de "Allianz"
ws5.append(["Aseguradora", "Fecha Reporte", 
            "Producto", "N° Póliza",
            "Prima($)","forma_pago","Tipo_comision"])
```

```python
for i in range(4000):
    # Aseguradora con errores de escritura
    aseguradora = random.choice(["Allianz", "Alianz", "Allian", "Allianz Seguros", "Allians"])

    # Fecha con posibles errores de formato
    if random.random() < 0.03:
        fecha = fake.date().replace("-", ".")  # formato raro
    else:
        fecha = fake.date_between(start_date='-6M', end_date='today').strftime("%d/%m/%Y")

    # Producto con opción inválida
    producto = random.choice(ramos + ["", "???", None])

    poliza = fake.bothify(text='??-#####')

    # Prima errónea
    if random.random() < 0.02:
        prima = random.choice([0, -5000])
    else:
        prima = round(random.uniform(10000, 500000), 2)

    pct_comision = random.choice([5, 10, 12.5])

    # Comisión mal calculada
    if random.random() < 0.04:
        comision = round(random.uniform(1000, 25000), 2)
    else:
        comision = round(prima * pct_comision / 100, 2)

    # Forma de pago inválida
    forma_pago_valor = random.choice(forma_pago + ["", "Pago raro", None])

    # Tipo de comisión incorrecto
    tipo_comision_valor = random.choice(tipo_comision + ["", "Extra", None])

    fila = [aseguradora, fecha, producto, poliza, prima, comision, forma_pago_valor, tipo_comision_valor]
    ws5.append(fila)

    # Duplicados con 5% de probabilidad
    if random.random() < 0.05:
        ws5.append(fila)
```
